```python
import sys
from pathlib import Path
from sklearn.ensemble import IsolationForest

sys.path.insert(0, str(Path.cwd().parent))

from loganalyzer.embedding import load_embeddings
from loganalyzer.data import read_log_file
```

```python
root_dir = str(Path.cwd().parent)
```

```python
lines = read_log_file(root_dir + "/data/server.log.2026-07-16")
embeddings = load_embeddings(root_dir + "/data/server.log.2026-07-16.embeddings.npy")
```

```python
# contamination=0.01 indica que esperamos cerca de 1% de anomalias nos dados
model = IsolationForest(contamination=0.005, random_state=42)
predicoes = model.fit_predict(embeddings)
```

```python
anomalias_idx = [idx for idx, pred in enumerate(predicoes) if pred == -1]
anomalias = [lines[i] for i in anomalias_idx]
```
